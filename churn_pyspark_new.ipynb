Machine Learning

```python
from pyspark.ml.clustering import KMeans
from pyspark.ml.linalg import Vectors
from pyspark.ml.feature import VectorAssembler
```

```python
vec_assembler = VectorAssembler(inputCols=['Total day charge', 'Total eve charge', 'Total night charge'], outputCol="features")
```

```python
vcluster_df = vec_assembler.transform(df)
vcluster_df.show()
# Spark expects the data to be a single vector
```

```python
kmeans = KMeans().setK(3).setSeed(1)
```

```python
kmodel = kmeans.fit(vcluster_df) # by default it will look for a column named 'features'
```

```python
kmodel.clusterCenters()
```

```python
from pyspark.ml.classification import LogisticRegression
```

```python
vcluster_df = vcluster_df.withColumn("Churn", col("Churn").cast("int"))
```

```python
lr = LogisticRegression(featuresCol='features', labelCol="Churn")
```

```python
lrModel = lr.fit(vcluster_df)
```

```python
lrModel.summary.areaUnderROC
```
